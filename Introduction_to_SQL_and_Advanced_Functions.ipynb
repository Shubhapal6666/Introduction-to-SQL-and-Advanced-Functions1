# question 1: explain the fundamental differences between ddl, dml, and dql commands in sql.
# provide one example for each type of command.

# answer:

# ddl (data definition language):
# ddl commands are used to create, modify, or delete database objects such as tables.
# example:
# create table students (
#     studentid int,
#     studentname varchar(50)
# );

# dml (data manipulation language):
# dml commands are used to insert, update, or delete data from tables.
# example:
# insert into students values (1, 'alice');

# dql (data query language):
# dql commands are used to retrieve data from the database.
# example:
# select * from students;

# question 2: what is the purpose of sql constraints?
# name and describe three common types of constraints.

# answer:

# sql constraints are rules that ensure the accuracy and integrity of data in a table.

# primary key:
# uniquely identifies each record in a table.
# example: customerid in a customers table.

# foreign key:
# creates a relationship between two tables.
# example: categoryid in the products table references the categories table.

# not null:
# ensures that a column cannot contain null values.
# example: productname should always have a value.

# other common constraints include unique, check, and default.

# question 4: what is a common table expression (cte) in sql?
# what are its main benefits? provide a simple sql example.

# answer:

# a common table expression (cte) is a temporary result set that can be referenced
# within a select, insert, update, or delete statement.

# benefits:
# - improves query readability.
# - simplifies complex queries.
# - makes sql code easier to maintain.
# - can be reused within the same query.

# example:

# with expensive_products as (
#     select productname, price
#     from products
#     where price > 100
# )
# select * from expensive_products;

# question 5: describe the concept of sql normalization and its primary goals.
# explain 1nf, 2nf, and 3nf.

# answer:

# normalization is the process of organizing data in a database to reduce redundancy
# and improve data integrity.

# goals:
# - reduce duplicate data.
# - improve consistency.
# - maintain data integrity.
# - make database management easier.

# first normal form (1nf):
# each column contains atomic (single) values and there are no repeating groups.

# second normal form (2nf):
# the table is in 1nf and all non-key attributes depend on the entire primary key.

# third normal form (3nf):
# the table is in 2nf and there are no transitive dependencies.
# non-key attributes depend only on the primary key.

# question 5: describe the concept of sql normalization and its primary goals.
# briefly explain the first three normal forms (1nf, 2nf, 3nf).

# answer:

# normalization is the process of organizing data in a database to reduce data redundancy
# and improve data integrity.

# primary goals:
# - eliminate duplicate data.
# - improve data consistency.
# - reduce storage space.
# - maintain data integrity.

# first normal form (1nf):
# - each column contains only atomic values.
# - no repeating groups.

# second normal form (2nf):
# - table must be in 1nf.
# - every non-key attribute depends on the entire primary key.

# third normal form (3nf):
# - table must be in 2nf.
# - no transitive dependency.
# - non-key columns depend only on the primary key.

# question 7:
# generate a report showing customername, email, and totalnumberoforders for each customer.
# include customers who have not placed any orders.

# answer:

# select
# c.customername,
# c.email,
# count(o.orderid) as totalnumberoforders
# from customers c
# left join orders o
# on c.customerid = o.customerid
# group by c.customerid, c.customername, c.email
# order by c.customername;

# question 9:
# use a common table expression (cte) and row_number() to display
# the top 2 most expensive products in each category.

# answer:

# with rankedproducts as (
# select
# c.categoryname,
# p.productname,
# p.price,
# row_number() over(
# partition by c.categoryname
# order by p.price desc
# ) as rn
# from products p
# join categories c
# on p.categoryid = c.categoryid
# )

# select
# categoryname,
# productname,
# price
# from rankedproducts
# where rn <= 2;
